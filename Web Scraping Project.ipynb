```python
import requests
from bs4 import BeautifulSoup
```

```python
#Step1:- Website to Scrape
web_url = "https://www.carwale.com/hyundai-cars/"
```

```python
#Step2:- Send a Get request to website.
response = requests.get(web_url)
response
```

```python
#Step3:- Helps you extract data from HTML easily.
soup = BeautifulSoup(response.text)
soup
```

```python
#Step4:- It is used to formats the HTML code into a clean, readable, and indented structure.
print(soup.prettify())
```

```python
#Step5:- To Get the image from server.
image_urls = []
for img in soup.find_all("img" , class_= "o-D o-hP o-iB o-iT o-G o-Q o-a0 o-aa o-hl o-ed"):
    image_urls.append(img["src"])

image_urls
```

```python
#Step6:- To Get the product names from server.
product_names = []
for tag in soup.find_all("h3", class_="o-j4 o-jq o-hM o-c4 o-jK"):
    product_names.append(tag.text.strip())
for n in product_names:
    print(n)
```

```python
#Step7:- To Get the prices from the server.
prices = []
price_tags = soup.find_all("span", class_="o-jJ o-jr o-j2 o-kJ")
for p in price_tags:
    prices.append(p.text.strip())
for price in prices:
    print(price)
```

```python
#Step8:-Prints data in a table-like format.
for name, image, price in zip(product_names, image_urls, prices):
    print("|", name, "|", image, "|", price, "|")
```

```python
#Step9:- To Download the File.
import csv
from IPython.display import FileLink
filename = "products.csv"
with open(filename, "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f, delimiter=',', quotechar='"', quoting=csv.QUOTE_ALL)
    writer.writerow(["Product Name", "Image URL", "Price"])
    for name, image, price in zip(product_names, image_urls, prices):
      writer.writerow([name, image, price])
print(f"CSV file '{filename}' created successfully!")
FileLink(filename)
```
